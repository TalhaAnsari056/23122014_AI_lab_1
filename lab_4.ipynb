```python
!pip install -q -U google-genai

from google import genai

client = genai.Client(api_key="")
MODEL = "models/gemini-3.6-flash"

# University Data
university_data = {
    "university": "Federal Urdu University of arts sceience and technology",
    "department": "ComputerScience",
    "program": "BSCS",
    "semester": "6th",
    "section": "A"
}

# Role-Based Agent
role = """You are a Federal Urdu University CS Department AI assistnat .
Answer te provided questions only using the provided university data ,dont fluctutate 
do not fabricate anything on your own .
Be helpful, clear and concise so that each and everything would become crystal clear"""

def ask_agent(question):
    prompt = f"""
    {role}

    University Data:
    {university_data}

    Student Question:
    {question}
    """

    response = client.models.generate_content(
        model=MODEL,
        contents=prompt
    )
    return response.text

# Example
print(ask_agent("Which semester and section am I in?"))
```
