```python
from transformers import pipeline

agent = pipeline("text-generation", model="Qwen/Qwen3-0.6B", device_map="auto")

def travel_agent(request):
    prompt = f"""You are a helpful Travel Agent.
Plan a practical trip based on the user's requirements.
Include destination, itinerary, estimated budget and useful tips.
User request: {request}
Give a concise and well-structured answer."""
    response = agent(prompt, max_new_tokens=400, do_sample=True,
                     temperature=0.7)
    return response[0]["generated_text"]

print(travel_agent("Plan a 5-day trip to Istanbul from Karachi with a budget of $1000."))
```
